```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
data = pd.read_csv("Iris.csv")
```

```python
data.sample(5)
```

```python
data.drop("Id",axis=1,inplace=True)
```

```python
# convert 3 species into 0, 1, 2

le = LabelEncoder()
data['Species'] = le.fit_transform(data['Species'])
```

```python
data.sample(5)
```

```python
x = data.drop("Species",axis=1)
y = data["Species"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.5,random_state=42)
```

```python
# Logistic Regression :

Logistic_Reg = LogisticRegression(max_iter=1000)
Logistic_Reg.fit(x_train,y_train)
```

```python
# Naive Baeyes :

NB = GaussianNB()
NB.fit(x_train,y_train)
```

```python
# KNN :

KNN = KNeighborsClassifier(n_neighbors=5)
KNN.fit(x_train,y_train)
```

```python
# Models Dictionary :

dic = {
    "Logistic Regression" : Logistic_Reg,
    "Naive Bayes (Gassian)" : NB,
    "KNN" : KNN
}

model_name=""
best_accuracy=0

for name,model in dic.items():
    print(f"\n{name} :")
    y_pred = model.predict(x)
    print("Accuracy:", accuracy_score(y, y_pred))
    if(accuracy_score(y, y_pred)>best_accuracy):
        best_accuracy = accuracy_score(y, y_pred)
        model_name=name
    print("cm:", confusion_matrix(y, y_pred))
    print("classification report:", classification_report(y, y_pred))
```

```python
print("Best Accuracy Model : ",model_name)
print("Accuracy : ",best_accuracy)
```
