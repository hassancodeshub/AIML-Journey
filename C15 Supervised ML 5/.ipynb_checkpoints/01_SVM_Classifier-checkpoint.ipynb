```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df = datasets.load_iris(as_frame = True).frame
```

```python
df.isnull().sum()
```

```python
df.sample(3)
```

```python
X = df.drop("target",axis = 1)
y = df["target"]
```

```python
X.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.3, random_state = 42, stratify = y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled  = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# Model Building :

svc = SVC()
svc.fit(x_train_scaled,y_train)

y_predict = svc.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("accuracy : ", accuracy_score(y_test,y_predict))
print("classification report : \n\n",classification_report(y_test,y_predict))
```

```python
# linear Kernel :

svc = SVC(kernel = 'linear')
svc.fit(x_train_scaled,y_train)

y_predict = svc.predict(x_test_scaled)

print("accuracy : ", accuracy_score(y_test,y_predict))
```

```python
# poly kernel :

svc = SVC(kernel = 'poly')
svc.fit(x_train_scaled,y_train)

y_predict = svc.predict(x_test_scaled)

print("accuracy : ", accuracy_score(y_test,y_predict))
```

```python
# sigmoid kernel :

svc = SVC(kernel = 'sigmoid')
svc.fit(x_train_scaled,y_train)

y_predict = svc.predict(x_test_scaled)

print("accuracy : ", accuracy_score(y_test,y_predict))
```

```python
# Hyperparameter :

c_vals = [0.5,1,2,3,4,5]

for c_val in c_vals:
    svc = SVC(kernel = 'rbf',C = c_val)
    svc.fit(x_train_scaled,y_train)
    
    y_predict = svc.predict(x_test_scaled)
    
    print("C : ",c_val," accuracy : ", accuracy_score(y_test,y_predict))
```
