```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
# Model Logic :

class KNNRegressor:
    def __init__(self,k=3):
        self.k = k

    def fit(self,x,y):
        self.x_train = x
        self.y_train = y

    def distance(self,x1,x2):
        return np.sqrt(np.sum((x2-x1)**2))
    
    def __predict_one(self,x):
        
        distances = [self.distance(x,x_train) for x_train in self.x_train]
        
        # Sort & Get the index :
        knn_index = np.argsort(distances)[:self.k]

        # Get the value from the y :
        y_train = [self.y_train[i] for i in knn_index]
        
        average = np.mean(y_train)

        return average

    def predict(self,x):
        y_pred = [self.__predict_one(e) for e in x]
        return np.array(y_pred)
```

```python
# Training data
X_train = np.array([[1], [2], [3], [4], [5]])
y_train = np.array([2, 4, 6, 8, 10])

# Test data
X_test = np.array([[1.5], [3.5], [6]])

# Create and train model
knn = KNNRegressor(k=3)
knn.fit(X_train, y_train)

# Predict
y_pred = knn.predict(X_test)
print("Predictions:", y_pred)
```
