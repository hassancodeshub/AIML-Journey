```python
import numpy as np
```

```python
class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weight = None

    def fit(self,x,y):
        m,n = x.shape

        x_b = np.c_[np.ones((m,1)),x]
        theta = ((np.linalg.inv(x_b.T@x_b))@(x_b.T@y))

        self.bias = theta[0]
        self.weight = theta[1:]

        y_pred = self.bias + np.dot(x,self.weight)

    def predict(self,x):
        y_pred = self.bias + np.dot(x,self.weight)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weight)
```
