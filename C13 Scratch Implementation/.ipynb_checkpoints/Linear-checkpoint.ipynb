```python
import numpy as np
```

```python
# Linear Regression :

class LinearRegression:
    def __init__(self,learning_rate=0.01,n_iter=1000):
        self.bias = None
        self.weight = None
        self.learning_rate = learning_rate
        self.n_iter = n_iter

    def fit(self,x,y):

        # Step 1 : Dimension
        m,n = x.shape # (no. of data points,no. features)

        # Step 2 : Initilization
        self.bias = 0
        self.weight = np.zeros(n) # [0,0,0,0,...,0] (r,1)

        for i in range(self.n_iter):
            # Step 3 : Equation
            y_pred = self.bias + np.dot(x,self.weight)
    
            # Step 4 : Derivatives
            db = (1 / m) * np.sum(y_pred - y) # Aggregate Sum
            dw = (1 / m) * np.dot(x.T,y_pred-y)
    
            # Step 5 : Weight Change
            self.bias -= self.learning_rate * db
            self.weight -= self.learning_rate * dw

    def predict(self,x):
        y_pred = self.bias + np.dot(x,self.weight)
        return y_pred
```

```python
# Testing Time :

x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(x, y)

y_pred = model.predict(x)
print(y_pred)

print(model.bias)
print(model.weight)
```
