```python
import numpy as np
```

```python
class LogisticRegression:
    
    def __init__(self,max_iter = 1000,learning_rate = 0.1):
        self.max_iter = max_iter
        self.learning_rate = learning_rate
        self.bias = None
        self.weight = None

    def get_probability(self,z):
        prob = z >= 0.5
        prob = prob.astype(int)
        return prob

    def sigmoid(self,z):
        return (1 / (1 + np.exp(-z)))
    
    def fit(self,x,y):
        m,n = x.shape

        self.bias = 0
        self.weight = np.zeros(n)

        
        for i in range(self.max_iter):

            z = self.bias + np.dot(x,self.weight)
            y_pred = self.sigmoid(z)
            
            db = (1/m) * (y_pred-y)
            dw = (1/m) * (np.dot(x.T,(y_pred-y)))

            self.bias -= self.learning_rate*db
            self.weight -= self.learning_rate*dw

    def predict(self,x):
        z = self.bias + np.dot(x,self.weight)
        y_pred = self.sigmoid(z)
        return self.get_probability(y_pred)

    def prob(self,x):
        z = self.bias + np.dot(x,self.weight)
        y_pred = self.sigmoid(z)
        print(y_pred)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

model.prob(X)
```
