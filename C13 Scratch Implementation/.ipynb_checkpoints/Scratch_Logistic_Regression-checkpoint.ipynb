```python
import numpy as np
```

```python
# Logistic Regression :

class LogisticRegression:
    def __init__(self,learning_rate=0.1,n_iter=1000):
        self.bias = None
        self.weight = None
        self.learning_rate = learning_rate
        self.n_iter = n_iter

    def __sigmoid(self,z):
        return (1 / (1 + np.exp(-z)))

    def fit(self,x,y):

        # Step 1 : Dimension
        m,n = x.shape # (no. of data points,no. features)

        # Step 2 : Initilization
        self.bias = 0
        self.weight = np.zeros(n) # [0,0,0,0,...,0] (r,1)

        for i in range(self.n_iter):
            # Step 3 : Equation
            z = self.bias + np.dot(x,self.weight)
            y_pred = self.__sigmoid(z)
            
            # Step 4 : Derivatives
            db = (1 / m) * (y_pred - y)
            dw = (1 / m) * np.dot(x.T,(y_pred-y))
    
            # Step 5 : Weight Change
            self.bias -= self.learning_rate * db
            self.weight -= self.learning_rate * dw

    def get_probabilities(self,x):
        z = self.bias + np.dot(x,self.weight)
        return self.__sigmoid(z)
    
    def predict(self,x,threshold = 0.5):
        prob = self.get_probabilities(x)
        y_pred_bool = prob>=threshold
        return y_pred_bool.astype(int)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.get_probabilities(X))
```
