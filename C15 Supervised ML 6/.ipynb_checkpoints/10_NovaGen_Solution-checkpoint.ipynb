```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,accuracy_score 
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import StackingClassifier

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

```python
data = pd.read_csv("novagen_dataset.csv")

data.info()
```

```python
X = data.drop("Target",axis=1)
y = data["Target"]

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, stratify=y,random_state=42)

std = StandardScaler()

cols = X.columns

X_train = pd.DataFrame(std.fit_transform(X_train),columns=cols)
X_test = pd.DataFrame(std.transform(X_test),columns=cols)
```

```python
#  Decision Tree Classifer :

dtc = DecisionTreeClassifier()

dtc.fit(X_train,y_train)

y_pred = dtc.predict(X_test)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
```

```python
#  LogisticRegression Classifer :

lr = LogisticRegression()

lr.fit(X_train,y_train)

y_pred = lr.predict(X_test)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
```

```python
# KNeighborsClassifier :

knn = KNeighborsClassifier()

knn.fit(X_train,y_train)

y_pred = knn.predict(X_test)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
```

```python
# Ensemble Learning - Stacking :

stack_class = StackingClassifier(
    estimators = [
        ("lr",lr),
        ("dtc",dtc),
        ("knn",knn)
    ],
    final_estimator = dtc,
    cv = 5
)

stack_class.fit(X_train,y_train)

y_pred = stack_class.predict(X_test)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
```

```python
print(classification_report(y_test,y_pred))
```
